<h3> PyTorch: Shallow Copy vs Deep Copy</h3>

<table border="1" style="border-collapse: collapse; width: 100%;">
  <tr style="background-color: #f2f2f2;">
    <th style="padding: 8px;">বিষয়</th>

    <th style="padding: 8px;">Deep Copy (ডিপ কপি)</th>
  </tr>
  <tr>
    <td style="padding: 8px;"><b>কীভাবে তৈরি হয়?</b></td>

  </tr>
  <tr>
    <td style="padding: 8px;"><b>মেমরি (RAM/GPU)</b></td>
    <td style="padding: 8px;">নতুন কোনো মেমরি নেয় না (একই মেমরি শেয়ার করে)</td>
    <td style="padding: 8px;">একদম নতুন মেমরি তৈরি করে</td>
  </tr>
  <tr>
    <td style="padding: 8px;"><b>স্বাধীনতা (Independence)</b></td>
    <td style="padding: 8px;">স্বাধীন নয় (একটিতে পরিবর্তন করলে অন্যটিতেও প্রভাব পড়ে)</td>
    <td style="padding: 8px;">সম্পূর্ণ স্বাধীন (একটির পরিবর্তনে অন্যটিতে কোনো প্রভাব পড়ে না)</td>
  </tr>

    <td style="padding: 8px;">Computation Graph শেয়ার করে</td>
    <td style="padding: 8px;"><code>.clone()</code> দিলে Computation Graph বজায় থাকে এবং Gradient ঠিকমতো Flow হতে পারে</td>
  </tr>
</table>

```python

# 1. Shallow Copy Test
a = torch.tensor([1, 2, 3])
b = a  # Shallow copy
b[0] = 999
print("Shallow Copy Output -> a:", a) # a-ও বদলে যাবে

# 2. Deep Copy Test
x = torch.tensor([1, 2, 3])
y = x.clone()  # Deep copy
y[0] = 999
print("Deep Copy Output -> x:", x) # x অপরিবর্তিত থাকবে
```
